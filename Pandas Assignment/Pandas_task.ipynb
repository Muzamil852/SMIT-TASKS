<h2>Task 11: Pandas Pivot Table Analysis
</h2>

```python
# This pivot table shows how High, Moderate, and Low Adaptivity are distributed among different education levels. It helps identify which education 
# level has the highest or lowest adaptability.# percentage demonstrates better adaptability
dataset.pivot_table(index="Education Level",columns="Adaptivity Level",aggfunc="size")
```

```python
# This pivot table compares students' financial conditions with their adaptability levels. It helps determine whether financial condition appears to
# influence students' ability to adapt to online learning.
dataset.pivot_table(index="Financial Condition",columns="Adaptivity Level",aggfunc="size")
```

```python
# This pivot table shows the distribution of Adaptivity Levels among students using different devices. It helps identify which device category has 
# the highest High Adaptivity and which has the lowest.
dataset.pivot_table(index="Device",columns="Adaptivity Level",aggfunc="size")
```
